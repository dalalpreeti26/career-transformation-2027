Dictionary in Python

```python
dict = {
        "Name" : "Preeti",
        "cgpa" : 9.6,
        "marks" : [98,97,95],
        "subjects": ('python','C','Java')
        }
print(dict)
type(dict)
print(dict['Name'] )  # to get the values using key
print(dict['subjects'])
dict['Name'] = 'Preeti Dalal'  # changes/replaced the name    #old value overwrites
print(dict)

dict['City'] = "Gurugram"  #added new key and value
print(dict)

null_dict = {}
print(null_dict)

null_dict['Name'] = "Preeti Dalal"
print(null_dict)
```

```python
#Nested Dictionary   # value of a parcticular key is stored as a dictionary

student = {
    'name' : 'Rahul Kumar',
    'subject' : {
        "maths" : 99,
        "chemistry" : 90,
        "Biology" : 95},
    "class" : 12

        }

print(student)
student['subject']['maths']

print(student.keys())
print(student.values())

#print(len(list(student.keys())) )  #Typecasting

student.items()

#pairs = list(student.items())
#print(pairs)

print(student["name"])
print(student.get("name"))


#print(student["name2"])   # will give error
print(student.get("name2"))   # wil give none  .get is method

student.update({"State": "Haryana"})
print(student)

new_dict = {"city": "Gurugram"}
student.update(new_dict)
print(student)
```
